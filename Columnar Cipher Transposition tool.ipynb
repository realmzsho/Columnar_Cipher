## Columnar Transposition tool

```python
from math import ceil
class transpositioncipher(object):
    def __init__(self,key): # assigning the key
        self.key=key.lower()
        self.order=self.get_ordered_list(self.key) #this is the order of key letters

    def get_ordered_list(self,key): # turn the key string to a numeric ordered value
        alphabet='abcdefghijklmnopqrstuvwxyz'
        # keeping the indexes of key letters in a list
        indexes_list=[] 
        for letter in key:
            indexes_list.append(alphabet.find(letter))
            
        # knowing the order of the key letters indexes
        order_list=[]
        for i in range(len(key)):
            index=len(key)-1
            for j in range(len(key)):
                if indexes_list[i]<indexes_list[j]: #this will know the order of the element
                    index-=1
            order_list.append(index)
        return order_list
                
    
    def encrypt_message(self,message):
        encrypted_message=''
        encryption_list=[]
        #initialize the encryption list
        for i in range(len(self.order)):
            encryption_list.append('')

        #making the distriution of letters due to order (encrypting)
        for i in range(len(message)):
            encryption_list[i%len(self.order)]=encryption_list[i%len(self.order)]+message[i]

        #setting the encryption message in string in order
        for i in range(len(self.order)):
            index=self.order.index(i)
            encrypted_message=encrypted_message+encryption_list[index]

        return encrypted_message

    
    def decrypt_message(self,message):
        decrypted_message=''
        decryption_list=[]
        #initialize the decryption_list
        for i in range(len(self.order)):
            decryption_list.append('')

        #finiding out how many rows are completed
        completed_rows=len(message)//len(self.order)
        #there may exist only one incomplete row
        #finding how many elements in this incompleted row
        rest=len(message)-completed_rows*len(self.order)
        #how many letters in the completed rows
        max_no_letters=ceil(len(message)/len(self.order))
        
        #put the characters in its column
        #الفكرة هنا لازم تعرف ايه العمود اللي فيه حروف اقل و ايه العمود المليان حروف
        letter_iterator=0
        for i in range(len(self.order)):
            order_index=self.order.index(i)
            if order_index<rest:      #هل العمود ده علي الناحية اللي مش ناقصها حرف في الاخر
                no_letters=max_no_letters
                decryption_list[order_index]=decryption_list[order_index]+message[letter_iterator:letter_iterator+no_letters]
                letter_iterator+=no_letters
            else:  #هل العمود ده علي الناحية اللي ناقصها حرف في اخر صف
                no_letters=completed_rows
                decryption_list[order_index]=decryption_list[order_index]+message[letter_iterator:letter_iterator+no_letters]+'#'  # ال# عشان املا مكان اخر حرف مش مكتوب
                letter_iterator+=no_letters
        
        #getting the decrypted text
        for i in range(max_no_letters):
            for j in range(len(decryption_list)):
                decrypted_message=decrypted_message+decryption_list[j][i]
        decrypted_message=decrypted_message.split('#')[0]
        
        return decrypted_message
```

```python
#### user interface

key = input("Enter the key for your cipher: ")
user_tool = transpositioncipher(key)
while (True):
    print('''\nWhat do you want to do?
    1) encrypt a message
    2) decrypt a message
    3) chenge the key
    4) exit''')
    choice = input('choose  1 , 2 , 3 or 4 : ')
    
    if choice =='1':
        text= input('Enter the message you want to encrypt : ')
        encrypted_text = user_tool.encrypt_message(text)
        print(f'encypted message is : {encrypted_text}')
        
        save=input('Want to save this in a file [y\\n]')
        if save == 'y':
            with open ('cipher.txt' , 'a') as f:
                f.writelines(encrypted_text + '\n')
            print('saved sucssesfully')
    
    elif choice == '2' :
        text= input('Enter the message you want to decrypt : ')
        decrypted_text = user_tool.decrypt_message(text)
        print(f'plain-text message is : {decrypted_text}')
        save=input('Want to save this in a file [y\\n]')
        if save == 'y':
            with open ('cipher.txt' , 'a') as f:
                f.writelines(decrypted_text +'\n')
            print('saved sucssesfully')
            

    elif choice == '3' : 
        key = input("Enter the new key : ")
        user_tool = transpositioncipher(key)
        print("key changed sucessfully")
    
    elif choice =='4' : 
        print("See you then")
        break

    else : print('Wrong input , try again')
        
    print('')
    print('='*50)
    print('')

    
```
